```python
import pikepdf
from google.colab import files
from pathlib import Path
import io

print("📂 Upload your password-protected or restricted PDF file:")
uploaded = files.upload()   # returns dict: filename -> content bytes

if not uploaded:
    print("No file uploaded. Exiting.")
else:
    # Take the first uploaded file
    input_filename = next(iter(uploaded))
    pdf_bytes = uploaded[input_filename]

    # 2. Ask for the password (leave blank if there is none)
    password = input("Enter PDF password (or press Enter if there is no open password): ")

    # Define output file name (you can change the path as needed)
    output_filename = "unlocked_" + input_filename
    output_path = Path("/content") / output_filename

    try:
        # 3. Open and remove security
        # pikepdf accepts bytes directly via io.BytesIO, or a file path.
        # We'll wrap the bytes in a BytesIO object.
        pdf_stream = io.BytesIO(pdf_bytes)
        pdf = pikepdf.open(pdf_stream, password=password)

        # Save the unlocked PDF
        pdf.save(str(output_path))
        print(f"✅ Success! Unlocked PDF saved as: {output_filename}")

        # 4. Download the resulting file
        files.download(str(output_path))

    except pikepdf.PasswordError:
        print("❌ Error: The password you entered is incorrect.")
    except Exception as e:
        print(f"❌ An error occurred: {e}")
```
